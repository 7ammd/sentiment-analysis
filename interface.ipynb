```python
import ipywidgets as widgets
from IPython.display import clear_output, display

# --- 1. CHATGPT-INSPIRED STYLING ---
chatgpt_style = widgets.HTML(
    value="""
    <style>
    .chat-container {
        background-color: #343541;
        color: #ECECF1;
        padding: 20px;
        border-radius: 12px;
        font-family: 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif;
        box-shadow: 0 4px 15px rgba(0,0,0,0.3);
    }
    .chat-header {
        font-size: 20px;
        font-weight: 600;
        margin-bottom: 15px;
        color: #fff;
        border-bottom: 1px solid #4d4d4f;
        padding-bottom: 10px;
    }
    </style>
    <div class="chat-container">
        <div class="chat-header">🤖 Sentiment Analysis Assistant</div>
    </div>
"""
)

# --- 2. WIDGET COMPONENTS ---
text_input = widgets.Textarea(
    value="",
    placeholder="Message Sentiment AI (e.g., 'I am really excited about this!')...",
    layout=widgets.Layout(
        width="100%", height="80px", margin="10px 0px 10px 0px"
    ),
)

model_dropdown = widgets.Dropdown(
    options=["Logistic Regression", "Naive Bayes", "Neural Network"],
    value="Logistic Regression",
    description="<b>Model:</b>",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="300px", margin="5px 0px 10px 0px"),
)

analyze_button = widgets.Button(
    description="Send 🚀", layout=widgets.Layout(width="120px", height="38px")
)
analyze_button.style.button_color = "#10a37f"
analyze_button.style.font_weight = "bold"

output_area = widgets.Output()


# --- 3. EVENT HANDLER (Directly calls your model functions) ---
def on_button_clicked(b):
  with output_area:
    clear_output()
    user_text = text_input.value.strip()

    if not user_text:
      print("❌ Please enter a message first.")
      return

    selected_model = model_dropdown.value

    try:
      # Directly calls your model function which already prints its own sentiment
      if selected_model == "Logistic Regression":
        logistic_regression(user_text)

      elif selected_model == "Naive Bayes":
        naive_bayes(user_text)

      elif selected_model == "Neural Network":
        neural_nets(user_text)

    except Exception as e:
      print(
          f"⚠️ Error: Make sure you ran the `%run` cell for this model first.\nDetails: {e}"
      )


analyze_button.on_click(on_button_clicked)

# --- 4. RENDER INTERFACE ---
display(
    widgets.VBox(
        [
            chatgpt_style,
            text_input,
            widgets.HBox([model_dropdown, analyze_button]),
            output_area,
        ]
    )
)
```
